## 2. Estructura y Calidad de los Datos

Antes de realizar el análisis exploratorio, es fundamental entender la estructura de nuestras tablas y verificar la calidad de los datos (presencia de valores nulos, tipos de datos correctos, etc.).

A continuación, revisaremos la información general de cada tabla:

```python
# Función simple para imprimir la información clave de un DataFrame
def resumir_dataframe(df, nombre):
    print(f"--- Resumen de la tabla: {nombre} ---")
    print(f"Filas: {df.shape[0]}, Columnas: {df.shape[1]}")
    print("\nValores Nulos por columna:")
    print(df.isnull().sum())
    print("\nTipos de datos:")
    print(df.dtypes)
    print("="*40 + "\n")

# Ejecutar el resumen para las tres tablas
resumir_dataframe(df_repos, "REPOSITORIOS")
resumir_dataframe(df_archivos, "ARCHIVOS")
resumir_dataframe(df_frontmatter, "FRONTMATTER")
```
